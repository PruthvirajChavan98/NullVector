```python
from nullvector.domain import ParseRequest, ParserSettings
from nullvector.ingest import parse_document

# 1. Define where the PDF is
pdf_path = "/home/pruthvi/projects/NullVector/903000608.pdf"

# 2. Configure the parser (Deterministic Substrate)
# By default, it will attempt native extraction first.
# 1. Use the verified path from your dpkg check
settings = ParserSettings(
    tessdata_path="/usr/share/tesseract-ocr/5/tessdata",
    ocr_dpi=300
)

# 3. Create the Request
# The parse_run_id is used for idempotency and artifact isolation.
request = ParseRequest(
    source_path=pdf_path,
    parse_run_id="spec_v1_run",
    settings=settings
)

# 4. Execute the Substrate
# This builds the Page Ledger and extracts all artifacts.
manifest = parse_document(request)

print(f"Ingestion complete. Document ID: {manifest.document_id}")
print(f"Artifacts preserved at: {manifest.artifact_root}")
```
